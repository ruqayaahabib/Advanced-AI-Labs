**1. Load the dataset into a DataFrame and add readable target labels.**

```python
# Import the breast cancer dataset
from sklearn.datasets import load_breast_cancer

# Load the dataset as a DataFrame
data = load_breast_cancer(as_frame=True)

# Copy the complete dataset
df = data.frame.copy()

# Rename target column to diagnosis
df = df.rename(columns={"target": "diagnosis"})

# Add readable labels
# 0 = malignant, 1 = benign
target_name_map = {0: "malignant", 1: "benign"}
df["diagnosis_name"] = df["diagnosis"].map(target_name_map)

# Display the dataset
df.head()
```

**2. Display the first 10 rows, shape, columns, target counts, missing values, and descriptive statistics.**

```python
# Display the first 10 rows
print("First 10 rows:")
display(df.head(10))

# Display the number of rows and columns
print("\nShape:")
print(df.shape)

# Display all column names
print("\nColumns:")
print(df.columns.tolist())

# Display the number of cases in each diagnosis
print("\nTarget counts:")
print(df["diagnosis_name"].value_counts())

# Check missing values in each column
print("\nMissing values:")
print(df.isna().sum())

# Display descriptive statistics
print("\nDescriptive statistics:")
display(df.drop(columns=["diagnosis_name"]).describe().T)
```

**3. Create at least five engineered features using differences, ratios, interactions, or summaries.**

```python
# Create a copy of the dataset for feature engineering
df_fe = df.copy()

# Small value to prevent division by zero
eps = 1e-6


# 1. Difference between worst radius and mean radius
df_fe["radius_gap"] = (
    df_fe["worst radius"] - df_fe["mean radius"]
)


# 2. Difference between worst area and mean area
df_fe["area_gap"] = (
    df_fe["worst area"] - df_fe["mean area"]
)


# 3. Ratio of mean area to mean radius
df_fe["area_per_radius"] = (
    df_fe["mean area"] / (df_fe["mean radius"] + eps)
)


# 4. Ratio of mean perimeter to mean radius
df_fe["perimeter_per_radius"] = (
    df_fe["mean perimeter"] / (df_fe["mean radius"] + eps)
)


# 5. Interaction between compactness and concavity
df_fe["compactness_concavity_interaction"] = (
    df_fe["mean compactness"] * df_fe["mean concavity"]
)


# List of the five new features
new_features = [
    "radius_gap",
    "area_gap",
    "area_per_radius",
    "perimeter_per_radius",
    "compactness_concavity_interaction"
]

# Display the new features
df_fe[new_features].head()
```

**4. Create three filtered DataFrames, for example high-radius cases, high-texture cases, and benign-only
cases.**

```python
# 1. Cases with radius higher than the average radius
high_radius = df_fe[
    df_fe["mean radius"] > df_fe["mean radius"].mean()
]

# 2. Cases with texture higher than the average texture
high_texture = df_fe[
    df_fe["mean texture"] > df_fe["mean texture"].mean()
]

# 3. Only benign cases
benign_only = df_fe[
    df_fe["diagnosis_name"] == "benign"
]


# Display the number of rows in each filtered DataFrame
print("High-radius cases:", high_radius.shape[0])
print("High-texture cases:", high_texture.shape[0])
print("Benign cases:", benign_only.shape[0])
```

**8. Identify at least four highly correlated feature pairs and briefly explain redundancy.**

```python
# Calculate the correlation matrix
corr_matrix = df_fe[selected_for_heatmap].corr()

# Get highly correlated feature pairs
high_corr_pairs = []

for i in range(len(corr_matrix.columns)):
    for j in range(i):
        correlation = corr_matrix.iloc[i, j]

        if abs(correlation) > 0.90:
            high_corr_pairs.append((
                corr_matrix.columns[i],
                corr_matrix.columns[j],
                correlation
            ))

# Display the first 4 highly correlated pairs
high_corr_pairs[:4]
```



> These feature pairs are highly correlated, which means they contain similar information. This can cause redundancy because multiple features may represent the same underlying characteristics of the tumor.



**9. Use correlation with the target to rank features and select the top six features.**

```python
# Prepare the features and target
X_fe = df_fe.drop(columns=["diagnosis", "diagnosis_name"])
y = df_fe["diagnosis"]

# Calculate correlation between each feature and the target
# Sort features by the strongest absolute correlation
corr_with_target = X_fe.corrwith(y).sort_values(
    key=lambda s: s.abs(),
    ascending=False
)

# Select and display the top 6 features
top_6_corr = corr_with_target.head(6)

print("Top 6 features by correlation with diagnosis:")
print(top_6_corr)
```

**10. Use SelectKBest to select the top six features and compare the result with correlation ranking.**

```python
# Import SelectKBest and f_classif
from sklearn.feature_selection import SelectKBest, f_classif

# Select the top 6 features
selector = SelectKBest(score_func=f_classif, k=6)
selector.fit(X_fe, y)

# Get the selected features
selected_mask = selector.get_support()
selected_features = X_fe.columns[selected_mask]

# Display the selected features
print("Top 6 features selected by SelectKBest:")
print(selected_features.tolist())
```

```python
# Compare SelectKBest features with correlation features
print("Correlation Top 6:")
print(top_6_corr.index.tolist())

print("\nSelectKBest Top 6:")
print(selected_features.tolist())
```



> Both methods selected the same six features, showing that these features have a strong relationship with the diagnosis.







**11.1 Apply PCA after standardization**

**12. Create a cumulative explained-variance plot and report how many components are needed to explain
at least 90% of variance.**

```python
import numpy as np
# Apply PCA using all components
pca_full = PCA()
X_pca_full = pca_full.fit_transform(X_scaled)

# Calculate explained variance
explained_variance = pca_full.explained_variance_ratio_

# Calculate cumulative explained variance
cumulative_variance = np.cumsum(explained_variance)

# Plot cumulative explained variance
plt.figure(figsize=(8, 4))

plt.plot(
    range(1, len(explained_variance) + 1),
    cumulative_variance,
    marker="o"
)

# Add a line at 90%
plt.axhline(0.90, linestyle="--")

plt.title("Cumulative Explained Variance by PCA Components")
plt.xlabel("Number of components")
plt.ylabel("Cumulative explained variance")

plt.show()

# Find number of components needed for at least 90%
components_needed_90 = np.argmax(cumulative_variance >= 0.90) + 1

print(
    "Number of components needed to explain at least 90% of variance:",
    components_needed_90
)
```

**13. Build one safe train-test pipeline using either selected features or PCA features, then report accuracy.**

```python
# Import everything needed for Q13
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_fe,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Create a safe pipeline
selected_model = Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("select", SelectKBest(score_func=f_classif, k=6)),
    ("model", LogisticRegression(max_iter=2000, random_state=42))
])

# Train the model using the training data
selected_model.fit(X_train, y_train)

# Make predictions using the testing data
y_pred_selected = selected_model.predict(X_test)

# Calculate and display the accuracy
accuracy = accuracy_score(y_test, y_pred_selected)

print("Accuracy using selected features:", accuracy)
```
